```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
'''
    === === ! === ===
    LOAD ALL THE DATA
    === === ! === ===
'''

''' YAG:Ce '''
# UV laser.
data = np.load('fitting_parameters_yagce_uvlaser_10-19-2026.npz')
est_yagce_uvlaser = data['arr_0']
err_yagce_uvlaser = data['arr_1']
# Micro-CT.
data = np.load('fitting_parameters_yagce_microct_10-19-2026.npz')
est_yagce_microct = data['arr_0']
err_yagce_microct = data['arr_1']
# Synchrotron.
data = np.load('fitting_parameters_yagce_synchrotron_10-19-2026.npz')
est_yagce_synchrotron = data['arr_0']
err_yagce_synchrotron = data['arr_1']
nrg_yagce_synchrotron = data['arr_2']

''' YAG:Yb '''
# Micro-CT.
data = np.load('fitting_parameters_yagyb_microct_10-19-2026.npz')
est_yagyb_microct = data['arr_0']
err_yagyb_microct = data['arr_1']
# Synchrotron.
data = np.load('fitting_parameters_yagyb_synchrotron_10-19-2026.npz')
est_yagyb_synchrotron = data['arr_0']
err_yagyb_synchrotron = data['arr_1']
nrg_yagyb_synchrotron = data['arr_2']
```

```python
# Set font globally.
plt.rcParams['pdf.fonttype'] = 42  # Type 42 (TrueType), required for Adobe Illustrator.
plt.rcParams['ps.fonttype'] = 42   # Ensures fonts are saved as TrueType in EPS.
plt.rcParams['font.family'] = 'Arial'
params = {'mathtext.default': 'regular' }          
plt.rcParams.update(params)

fig, axes = plt.subplots(1, 3, figsize=(1.75, 0.75), sharey=True, gridspec_kw={'width_ratios': [1, 1, 3]})

''' UV laser. '''
ax = axes[0]
ax.errorbar(
    1, est_yagce_uvlaser[0],
    yerr=err_yagce_uvlaser[0],
    marker='o', color='green', markersize=3
    )
ax.set_xticks([])
# ax.set_ylabel('λ (µm)', fontsize=9)
# ax.set_title('UV', fontsize=9)
ax.tick_params(axis='both', which='major', labelsize=7)

''' Micro-CT. '''
ax = axes[1]
ax.errorbar(
    1, est_yagce_microct[0],
    yerr=err_yagce_microct[0],
    marker='o', color='green', markersize=3
    )
ax.errorbar(
    1, est_yagyb_microct[0],
    yerr=err_yagyb_microct[0],
    marker='s', color='red', markersize=3
    )
ax.set_xticks([])
# ax.set_title('µ-CT', fontsize=9)

''' Synchrotron. '''
ax = axes[2]
ax.errorbar(
    nrg_yagce_synchrotron, est_yagce_synchrotron[:,0],
    yerr=err_yagce_synchrotron[:,0],
    marker='o', color='green', markersize=3
    )
ax.errorbar(
    nrg_yagyb_synchrotron, est_yagyb_synchrotron[:,0],
    yerr=err_yagyb_synchrotron[:,0],
    marker='s', color='red', markersize=3
    )
ax.set_xlim([8, 24])
ax.set_xticks([8, 12, 16, 20, 24])
ax.set_xticklabels([])
ax.set_ylim([1.5e2, 1.5e4])
# ax.set_xlabel('Energy (keV)', fontsize=9)
# ax.set_title('Synchrotron', fontsize=9)

ax.set_yscale('log')

plt.savefig("fig_3_lambda_fit.pdf", format="pdf")
```
